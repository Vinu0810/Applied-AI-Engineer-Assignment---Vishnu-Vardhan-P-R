```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import re
import ast
```

```python
#The Dataframes
```

```python
appstore_data_temp1 = pd.read_csv("dataset/appstore_apps_by_name_1.csv")
appstore_data_temp2 = pd.read_csv("dataset/appstore_apps_by_name_2.csv")
appstore_data_temp3 = pd.read_csv("dataset/appstore_apps_by_name_3.csv")
appstore_data_temp4 = pd.read_csv("dataset/appstore_apps_by_name_4.csv")
playstore_data = pd.read_csv("dataset/cleaned_googleplaystore.csv")
```

```python
#CHecking if the app names in the app store dataset are same as the play store dataset
```

```python
appstore_app_names_temp1 = appstore_data_temp1["App"].dropna().unique().tolist()
appstore_app_names_temp2 = appstore_data_temp2["App"].dropna().unique().tolist()
appstore_app_names_temp3 = appstore_data_temp3["App"].dropna().unique().tolist()
appstore_app_names_temp4 = appstore_data_temp4["App"].dropna().unique().tolist()
playstore_app_names = playstore_data["App"].dropna().unique().tolist()
```

```python
def check_app_names(temp):
    count = 0
    for i in temp:
        if i in playstore_app_names:
            count += 1
    print(len(temp), count)
```

```python
check_app_names(appstore_app_names_temp1)
check_app_names(appstore_app_names_temp2)
check_app_names(appstore_app_names_temp3)
check_app_names(appstore_app_names_temp4)
```

```python
#Combining the Dataframes of the appstore datasets
```

```python
appstore_data = pd.concat([appstore_data_temp1, appstore_data_temp2, appstore_data_temp3, appstore_data_temp4], ignore_index=True)
```

```python
appstore_data.info()
```

```python
#Dropping unncessary fields of the appstore data
```

```python
appstore_data = appstore_data.drop(columns=["as_title", "as_released", "as_updated", "as_current_version_score", "as_current_version_reviews"], axis = 1)
```

```python
appstore_data = appstore_data.drop(columns=["as_app_id"], axis = 1)
```

```python
appstore_data.info()
```

```python
print(playstore_data.isnull().sum())
```

```python
def rating_change(x):
    if isinstance(x, str):
        x = x.strip()
        return x.split(", ")
    return np.nan
```

```python
appstore_data["as_genres"] = appstore_data["as_genres"].apply(rating_change)
```

```python
#changing dtype of reviews column
```

```python
appstore_data["as_reviews"] = appstore_data["as_reviews"].astype(float)
```

```python
appstore_data.head()
```

```python
#Dropping unncessary columns in playstore dataset
```

```python
playstore_data = playstore_data.drop(columns=["curr_ver", "and_ver"], axis = 1)
```

```python
playstore_data.head()
```

```python
#Standardizing Content Rating in playstore data
```

```python
playstore_data["Content Rating"].value_counts()
```

```python
def stand_cont_rat(x):
    if isinstance(x, str):
        x = x.strip()
        if x == "Mature 17+":
            return "Mature"
        elif x == "Everyone 10+":
            return "Teen"
        elif x == "Adults only 18+":
            return "Adults"
        else:
            return x
    return np.nan
```

```python
playstore_data["Content Rating"] = playstore_data["Content Rating"].apply(stand_cont_rat)
```

```python
playstore_data["Content Rating"].value_counts()
```

```python
#Last Updated Column of Playstore Dataset
```

```python
"""Since the playstore dataset is very old and the appstore one is relatively upto date, the "Last Updated" values for the playstore data 
   are in 2018. Since there is no latest data for them, and considering we need a balance in that, we are going to avoid the appstore 
   "Last Updated" values. Consider only the playstore's value for the same."""
```

```python
#Size Column of Playstore Dataset
```

```python
"""Just like the Last Updated Updated field, Size also chnages through years. So im using the playstore data's Size just as it is."""
```

```python
#Adding store column to the dataset
```

```python
appstore_app_names = appstore_data["App"].dropna().unique().tolist()

print(len(appstore_app_names))
```

```python
playstore_data["Store"] = "[android]"
```

```python
for app in playstore_data["App"]:
    if app.strip() in appstore_app_names:
        playstore_data.loc[playstore_data["App"] == app, "Store"] = "[android, apple]"
```

```python
playstore_data["Store"].value_counts()
```

```python
#merging genre names in the datasets
```

```python
playstore_data["new_gen"] = "Yes"
```

```python
app_genres_dict = {}
for idx, row in appstore_data.iterrows():
    app_genres_dict[row["App"]] = ast.literal_eval(str(row["as_genres"]))

play_genres_dict = {}
for idx, row in playstore_data.iterrows():
    play_genres_dict[row["App"]] = ast.literal_eval(str(row["Genres"]))
```

```python
for app in appstore_app_names:
    for i in app_genres_dict[app]:
        if i not in play_genres_dict[app]:
            play_genres_dict[app].append(i)
```

```python
play_genres_dict
```

```python
for app in playstore_data["App"]:
    if app.strip() in appstore_app_names:
        playstore_data.loc[playstore_data["App"] == app, "new_gen"] = str(play_genres_dict[app])
```

```python
playstore_data = playstore_data.drop(["Genres"], axis = 1)
playstore_data = playstore_data.rename(columns={"new_gen" : "Genres"})
```

```python
#merging category names in the datasets
```

```python
playstore_data["new_cat"] = "Yes"
```

```python
app_cat_dict = {}
for idx, row in appstore_data.iterrows():
    app_cat_dict[row["App"]] = row["as_primary_genre"]

play_cat_dict = {}
for idx, row in playstore_data.iterrows():
    play_cat_dict[row["App"]] = ast.literal_eval(str(row["Category"]))
```

```python
for app in appstore_app_names:
    i = app_cat_dict[app]
    if i not in play_cat_dict[app]:
        play_cat_dict[app].append(i)
```

```python
for app in playstore_data["App"]:
    if app.strip() in appstore_app_names:
        playstore_data.loc[playstore_data["App"] == app, "new_cat"] = str(play_cat_dict[app])
```

```python
playstore_data = playstore_data.drop(["Category"], axis = 1)
playstore_data = playstore_data.rename(columns={"new_cat" : "Category"})
```

```python
#cresating a min_age field
```

```python
age_map = {
    "Everyone": 5,
    "Teen": 13,
    "Mature": 17,
    "Adults": 18,
    "Unrated": 0
}

playstore_data["Min Age"] = playstore_data["Content Rating"].map(age_map)
```

```python
playstore_data["Content Rating"].value_counts()
```

```python
#Handling the reviews
```

```python
median_reviews = appstore_data["as_reviews"].median()
print(median_reviews)
```

```python
playstore_data["new_rev"] = median_reviews
```

```python
for app in playstore_data["App"]:
    old = playstore_data.loc[playstore_data["App"] == app, "Reviews"].iloc[0]
    if app in appstore_app_names:
        temp = appstore_data.loc[appstore_data["App"] == app, "as_reviews"].iloc[0]
        new = old + temp
    else:
        new = old + median_reviews
    playstore_data.loc[playstore_data["App"] == app, "new_rev"] = new
```

```python
playstore_data = playstore_data.drop(["Reviews"], axis = 1)
playstore_data = playstore_data.rename(columns={"new_rev" : "Reviews"})
```

```python
#Handling the Price
```

```python
for app in playstore_data["App"]:
    old = playstore_data.loc[playstore_data["App"] == app, "Price"].iloc[0]
    if app in appstore_app_names:
        temp = appstore_data.loc[appstore_data["App"] == app, "as_price"].iloc[0]
        old = old + temp
    playstore_data.loc[playstore_data["App"] == app, "Price"] = old
```

```python
#Handling Type
```

```python
playstore_data["Type"].value_counts()
```

```python
for app in playstore_data["App"]:
    price = playstore_data.loc[playstore_data["App"] == app, "Price"].iloc[0]
    type = playstore_data.loc[playstore_data["App"] == app, "Type"].iloc[0]
    if price == 0:
        playstore_data.loc[playstore_data["App"] == app, "Type"] = "Free"
    else:
        playstore_data.loc[playstore_data["App"] == app, "Type"] = "Paid"
```

```python
#Handling score
```

```python
playstore_data["new_rat"] = 0.0
```

```python
for app in playstore_data["App"]:
    old = playstore_data.loc[playstore_data["App"] == app, "Rating"].iloc[0]
    if app in appstore_app_names:
        new = appstore_data.loc[appstore_data["App"] == app, "as_score"].iloc[0]
        old = (old + new) / 2
    old = round(old, 2)
    playstore_data.loc[playstore_data["App"] == app, "new_rat"] = old
```

```python
playstore_data = playstore_data.drop(["Rating"], axis = 1)
playstore_data = playstore_data.rename(columns={"new_rat" : "Rating"})
```
